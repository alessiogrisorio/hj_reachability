Isosuperfici a sinistra e tre confronti XY a destra. Le velocità sono fisse; theta_rel varia tra le righe. Parametri fissi da indicare nelle caption. Usa i file originali `eggert.npz` e `time_eggert.npz`, visualizzati come PRM e TBR. I dati sono locali: eseguire Restart Kernel / Run All per generare le anteprime.

## Configurazione

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "scripts" / "compare_terminal_values.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Apri il notebook nel repository hj_reachability.")
for folder in (ROOT, ROOT / "grafici"):
    if str(folder) not in sys.path:
        sys.path.insert(0, str(folder))

from plot_utils import set_style, export_figure, finish_figure
from scripts import compare_terminal_values as comparison

set_style()
DATA_DIRECTORY = ROOT / "results" / "brt"
FIELD_KEYS = ("V0", "BRT")
V_H_FIXED = 6.0
V_E_FIXED = 6.0
DELTA_E_FIXED = 0.0
THETA_SLICES = (0.0, np.pi / 6, np.pi / 3)
V0_THETA_LIMITS = (-np.pi / 2, np.pi / 2)
BRT_THETA_LIMITS = (-np.pi, np.pi)
SURFACE_SMOOTHING = 0.8
FIGURE_SIZE = (11.5, 7.0)
EXPORT = True
FORMATS = ("pdf",)
figures = {}
```

```python
def show_comparison(pair):
    comparison.DATA_DIRECTORY = DATA_DIRECTORY
    comparison.V_H = V_H_FIXED
    comparison.V_E_3D = V_E_FIXED
    comparison.V_E_SLICE = V_E_FIXED
    comparison.DELTA_E = DELTA_E_FIXED
    comparison.THETA_SLICES = tuple(THETA_SLICES)
    comparison.THETA_LIMITS_3D = V0_THETA_LIMITS
    comparison.BRT_THETA_LIMITS_3D = BRT_THETA_LIMITS
    comparison.SMOOTHING_SIGMA = SURFACE_SMOOTHING
    comparison.FIGURE_SIZE = FIGURE_SIZE
    for key in FIELD_KEYS:
        name = f"{pair[0]}_vs_{pair[1]}_{key.lower()}"
        fig = comparison.make_figure(pair, key, save=False, return_figure=True)
        finish_figure(fig)
        figures[name] = fig
        plt.show()
        if EXPORT:
            export_figure(fig, nome=name, cartella=f"confronti_{key.lower()}",
                          formati=FORMATS, sovrascrivi=True)
        plt.close(fig)
```

## Euclidean e DCE

```python
show_comparison(('euclidean', 'dce'))
```

## BTN e STN

```python
show_comparison(('btn', 'stn'))
```

```python
show_comparison(('btn', 'long_lat'))
```
